**4-5. Рассчитайте среднее значение и среднеквадратичное отклонение возраста тех, кто получает более 50K в год (признак *salary*) и тех, кто получает менее 50K в год**

**8. Среди кого больше доля зарабатывающих много (>50K): среди женатых или холостых мужчин (признак *marital-status*)? Женатыми считаем тех, у кого *marital-status* начинается с *Married* (Married-civ-spouse, Married-spouse-absent или Married-AF-spouse), остальных считаем холостыми.**

```python
married_men = data[
    (data['sex'] == 'Male') &
    (data['marital-status'].str.startswith('Married'))
]

high_salary_married_men = (married_men['salary'] == '>50K').mean()
high_salary_married_men
```

```python
single_men = data[
    (data['sex'] == 'Male') &
    (~data['marital-status'].str.startswith('Married'))
]

high_salary_single_men = (single_men['salary'] == '>50K').mean()
high_salary_single_men
```

```python
if high_salary_married_men > high_salary_single_men:
    print('The share is higher among married men.')
else:
    print('The share is higher among single men.')
```

**11.Сгруппируйте людей по возрастным группам *young*, *adult*, *retiree*, где:**
* *young* соответствует 16-35 лет
* *adult* - 35-70 лет
* *retiree* - 70-100 лет

**Проставьте название соответствуещей группы для каждого человека в новой колонке AgeGroup**

```python
condlist = [((data['age'] >= 16) & (data['age'] < 35)), ((data['age'] >= 35) & (data['age'] < 70)), ((data['age'] >= 70) & (data['age'] < 100))]
choicelist = ['young', 'adult', 'retiree']
data['AgeGroup'] = np.select(condlist, choicelist, default='unknown')
data.head()
```

**12-13. Определите количество зарабатывающих >50K в каждой из возрастных групп (колонка AgeGroup), а также выведите название возрастной группы, в которой чаще зарабатывают больше 50К (>50K)**

```python
age_groups_high_salary = data[data['salary']=='>50K'].groupby(by=['AgeGroup'])['salary'].count()
age_groups_high_salary
```

```python
print(age_groups_high_salary.idxmax())
```

**14. Сгруппируйте людей по типу занятости (колонка occupation) и определите количество людей в каждой группе. После чего напишите функцию фильтрации filter_func, которая будет возвращать только те группы, в которых средний возраст (колонка age) не больше 40 и в которых все работники отрабатывают более 5 часов в неделю (колонка hours-per-week)**

```python
occupations = data.groupby('occupation')
print(occupations.size())
```

```python
def filter_func(group):
    return (
        group['age'].mean() <= 40
        and
        (group['hours-per-week'] > 5).all()
    )
```

```python
filtered = (occupations.filter(filter_func).groupby('occupation').size())
filtered
```
